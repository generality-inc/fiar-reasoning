## Add outcome

```python
# model name aliases from reasoning_depth.ipynb
MODEL_NAME_ALIASES = {
    # DeepSeek models: vertex -> fireworks
    'deepseek-ai/deepseek-v3.1-maas': 'accounts/fireworks/models/deepseek-v3p1',
    'deepseek-ai/deepseek-r1-0528-maas': 'accounts/fireworks/models/deepseek-r1-0528',
    
    # Claude models: vertex-anthropic -> anthropic
    'claude-opus-4-1@20250805': 'claude-opus-4-1-20250805',
    'claude-sonnet-4@20250514': 'claude-sonnet-4-20250514',
}

def get_legacy_model_name(model_id: str) -> str:
    """
    Get the legacy/old model name for backward compatibility.
    """
    return MODEL_NAME_ALIASES.get(model_id, model_id)

def normalize_model_name(model_id: str) -> str:
    """
    Normalize model names to use legacy names for backward compatibility.
    """
    return get_legacy_model_name(model_id)

def get_model_identifier(player_data) -> str:
    """
    Get model identifier including reasoning effort if present.
    Normalizes to legacy names for backward compatibility.
    Handles special cases like deepseek-v3.1 thinking config.
    """
    model_name = player_data.get('model', '') or ''
    config = player_data.get('config', {}) or {}
    reasoning_effort = config.get('reasoning_effort', '')
    
    # Special case: deepseek-v3.1 with thinking config
    # Only the thinking:true version maps to old model
    if model_name == 'deepseek-ai/deepseek-v3.1-maas':
        thinking = config.get('thinking', False)
        if thinking:
            # Old thinking version
            normalized_model = 'accounts/fireworks/models/deepseek-v3p1'
            return f"{normalized_model}-medium"
        else:
            # NEW non-thinking version - keep as-is
            return model_name
    
    # Special case: deepseek-r1 vertex always maps to old with -medium
    if model_name == 'deepseek-ai/deepseek-r1-0528-maas':
        normalized_model = 'accounts/fireworks/models/deepseek-r1-0528'
        return f"{normalized_model}-medium"
    
    # Default: normalize + add -<reasoning_effort> if present
    normalized_model = normalize_model_name(model_name)
    if reasoning_effort:
        # avoid duplicate suffix
        if not str(normalized_model).endswith(f"-{reasoning_effort}"):
            return f"{normalized_model}-{reasoning_effort}"
    return normalized_model
```

```python
# collect all game results
game_results = []
# all_games = df['game_path'].unique()

for game_path in df['game_path']:
    full_path = './aws-logs/four_in_a_row-fen/' + game_path
    
    with open(full_path, 'r') as f:
        data = json.load(f)
    
    # Extract result and models
    result = data.get('result', {})
    model_pair_key = data.get('model_pair_key', '')
    players = data.get('players', [])
    
    # Get the two models
    model_0 = get_model_identifier(players[0])
    model_1 = get_model_identifier(players[1])
    
    # Determine outcome (from perspective of player 0)
    game_type = result.get('type', '')
    winner = result['metadata'].get('winner', None)
    
    if game_type == 'win' or game_type == 'forfeit':
        if winner == 0:
            outcome = 1.0  # Player 0 wins
        elif winner == 1:
            outcome = 0.0  # Player 1 wins
        else:
            outcome = 0.5  # Should not happen for type='win'
    elif game_type == 'draw':
        outcome = 0.5  # Draw
    else:
        raise ValueError('Unclear game outcome.')
    
    game_results.append({
        # 'game_path': game_path,
        'model_0': model_0,
        'model_1': model_1,
        'outcome': outcome  # 1.0 = model_0 wins, 0.0 = model_1 wins, 0.5 = draw
    })
```

```python
df_game_results = pd.DataFrame(game_results)
df_game_results

df['outcome'] = df_game_results['outcome']

# df = pd.concat([df, df_game_results], axis = 1)

df['outcome_perspective'] = np.where(
    df['model_names'] == df_game_results['model_0'],
    df['outcome'],
    1 - df['outcome']
)
```

```python
df_grouoped = df.groupby(['game_path', 'model_names'])['outcome_perspective'].mean().reset_index()
df_grouoped.groupby(['model_names'])['outcome_perspective'].mean().reset_index()
```

## Add number of tokens

```python
# function to get token count from a game event
def get_reasoning_tokens(game_path, turn_idx):
    """
    Get the total reasoning + output tokens for a specific turn.
    
    Args:
        game_path: Path to the game log file (relative)
        turn_idx: Index in the events list
    
    Returns:
        Dictionary with token counts or None if not found
    """
    
    full_path = Path('./aws-logs/four_in_a_row-fen') / game_path

    with open(full_path, 'r') as f:
        game_log = json.load(f)
    
    # get the event at this index
    events = game_log.get('events', [])
    if turn_idx >= len(events):
        raise ValueError('Turn index out of range.')
    
    event = events[turn_idx]
    
    # check if this is a turn event with generation data
    if event.get('type') != 'turn':
        raise ValueError('Not a turn.')
    
    generation = event.get('generation', {})
    usage = generation.get('usage', {})

    if usage:
        # extract token counts from various possible locations
        result = {
            'input_tokens': 0,
            'output_tokens': 0,
        }
        
        # try to get input tokens
        if 'input_tokens' in usage:
            result['input_tokens'] = usage['input_tokens']
        
        # try to get output tokens (reasoning + regular output)
        if 'output_tokens' in usage:
            result['output_tokens'] = usage['output_tokens']
    
    else:
        result = None
    
    return result

token_data = []

for idx, row in df.iterrows():
    game_path = row['game_path']
    turn_idx = row['turn_idx']
    
    tokens = get_reasoning_tokens(game_path, turn_idx)
    
    if tokens:
        token_data.append(tokens)
    else:
        # default values if we can't find the data
        token_data.append({
            'input_tokens': None,
            'output_tokens': None,
        })
```

```python
# convert to DataFrame and add to original
token_df = pd.DataFrame(token_data)

# add new columns
df['input_tokens'] = token_df['input_tokens']
df['output_tokens'] = token_df['output_tokens']
```
